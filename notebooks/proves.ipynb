```python
from ultralytics import YOLO
import cv2
import os
import yt_dlp
```

```python
path_principal = os.path.join(os.getcwd(), '../')
```

```python
path_principal
```

```python

model = YOLO(path_principal+'yolov8n.pt')
```

```python
video_url = 'https://www.youtube.com/watch?v=Vape2-zhwmI'  # Reemplaza por tu enlace
output_path = path_principal+'videos/derbi_it.mp4'

ydl_opts = {
    'format': 'bestvideo',
    'merge_output_format': 'mp4',
    'outtmpl': output_path
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([video_url])
```

```python
results = model.predict(
    source=path_principal+'clips/clip2.mp4',
    save=True,          # Guarda el vídeo con las detecciones
    conf=0.25,          # Umbral de confianza
    imgsz=640,          # Tamaño de entrada (opcional)
    device='cpu'        # Cambia a 'cuda' si usas GPU
)
```
